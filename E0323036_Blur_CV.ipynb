```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
img = cv2.imread('/content/test.jfif')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
#Blur
blur = cv2.blur(img_rgb, (5, 5))
#Box Filter
box_filter = cv2.boxFilter(img_rgb, -1, (5, 5))
#Gaussian Blur
gaussian_blur = cv2.GaussianBlur(img_rgb, (5, 5), 0)

#Median Blur
median_blur = cv2.medianBlur(img_rgb, 5)
#Laplacian
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
laplacian = cv2.Laplacian(gray_img, cv2.CV_64F)
laplacian = np.uint8(np.absolute(laplacian)) # Convert back to uint8

#Visualization
titles = ['Original', 'Blur', 'Box Filter', 'Gaussian Blur', 'Median Blur', 'Laplacian']
images = [img_rgb, blur, box_filter, gaussian_blur, median_blur, laplacian]

plt.figure(figsize=(16, 10))

for i in range(6):
    plt.subplot(2, 3, i+1)
    if titles[i] == 'Laplacian (Edges)':
        plt.imshow(images[i], cmap='gray')
    else:
        plt.imshow(images[i])
    plt.title(titles[i])
    plt.xticks([]), plt.yticks([])
plt.tight_layout()
plt.show()
```
