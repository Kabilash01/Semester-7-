```python
#EXP1-Convert it from RGB to HSV, HLS color space using OpenCV and Display the
#original and converted images side by side.
import cv2

import matplotlib.pyplot as plt

img_bgr = cv2.imread("/content/images.jpg")

# Convert BGR to RGB for display / RGB-based processing
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

# Convert RGB to HSV and HLS
img_hsv = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2HSV)
img_hls = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2HLS)

# Display original and converted images side by side
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(img_rgb)
axes[0].set_title("Original RGB")
axes[0].axis("off")

axes[1].imshow(img_hsv)
axes[1].set_title("HSV")
axes[1].axis("off")

axes[2].imshow(img_hls)
axes[2].set_title("HLS")
axes[2].axis("off")

plt.tight_layout()
plt.show()
```

```python
#EXP 6 -Image Manipulation: Use OpenCV functions to apply different image manipulations
#(e.g., rotation, flipping).

import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()

# Get uploaded file name
image_name = list(uploaded.keys())[0]
# Read image
image = cv2.imread(image_name)

# Rotate image (90 degrees clockwise)
rotated_image = cv2.rotate(image, cv2.ROTATE_90_CLOCKWISE)

# Flip image horizontally
flip_horizontal = cv2.flip(image, 1)

# Flip image vertically
flip_vertical = cv2.flip(image, 0)

# Convert BGR to RGB for display
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
rotated_rgb = cv2.cvtColor(rotated_image, cv2.COLOR_BGR2RGB)
horizontal_rgb = cv2.cvtColor(flip_horizontal, cv2.COLOR_BGR2RGB)
vertical_rgb = cv2.cvtColor(flip_vertical, cv2.COLOR_BGR2RGB)

# Display side by side
fig, axes = plt.subplots(1, 4, figsize=(20, 6))

axes[0].imshow(image_rgb)
axes[0].set_title("Original Image")
axes[0].axis("off")

axes[1].imshow(rotated_rgb)
axes[1].set_title("Rotated Image")
axes[1].axis("off")

axes[2].imshow(horizontal_rgb)
axes[2].set_title("Horizontally Flipped")
axes[2].axis("off")

axes[3].imshow(vertical_rgb)
axes[3].set_title("Vertically Flipped")
axes[3].axis("off")

plt.tight_layout()
plt.show()

# Save images
cv2.imwrite("rotated_image.jpg", rotated_image)
cv2.imwrite("horizontal_flip.jpg", flip_horizontal)
cv2.imwrite("vertical_flip.jpg", flip_vertical)
```

```python
#EXP2-Implement blending and overlaying images using OpenCV
import cv2
import matplotlib.pyplot as plt
from google.colab import files
from google.colab.patches import cv2_imshow

# 1. Upload and Load Images
print("Upload your overlay image (e.g., images.jpg):")
upload1 = files.upload()
img1_name = list(upload1.keys())[0]
img1 = cv2.imread(f"/content/{img1_name}")

print("Upload your background image (e.g., space.jpg):")
upload2 = files.upload()
img2_name = list(upload2.keys())[0]
img2 = cv2.imread(f"/content/{img2_name}")

# Check if images were loaded successfully
if img1 is None:
    print(f"Error: Could not load overlay image from /content/{img1_name}. Please ensure it's a valid image file.")
elif img2 is None:
    print(f"Error: Could not load background image from /content/{img2_name}. Please ensure it's a valid image file.")
else:
    # Display original images using Colab's native viewer
    print("Original Images:")
    cv2_imshow(img1)
    cv2_imshow(img2)

    # 2. Prepare for Overlay
    overlay = img1.copy()
    background = img2.copy()

    # Resize the overlay to fit target dimensions
    overlay_resized = cv2.resize(overlay, (250, 250))

    # Set top-left placement coordinates
    x, y = 100, 1

    # Get exact dimensions of the resized overlay
    h, w = overlay_resized.shape[:2]

    # 3. Perform Region of Interest (ROI) Slicing / Overlay
    background[y:y+h, x:x+w] = overlay_resized

    # 4. Perform Alpha Blending (Bonus Extension)
    # This blends the two images seamlessly instead of a hard replacement
    background_roi = img2[y:y+h, x:x+w]  # Extract the background patch
    resized_bg_roi = cv2.resize(background_roi, (250, 250)) # Ensure identical size

    # Blend formula: dst = src1 * alpha + src2 * beta + gamma
    blended_roi = cv2.addWeighted(src1=overlay_resized, alpha=0.6, src2=resized_bg_roi, beta=0.4, gamma=0)

    blended_background = img2.copy()
    blended_background[y:y+h, x:x+w] = blended_roi

    # 5. Convert BGR to RGB for Matplotlib Display
    background_rgb = cv2.cvtColor(background, cv2.COLOR_BGR2RGB)
    blended_rgb = cv2.cvtColor(blended_background, cv2.COLOR_BGR2RGB)

    # 6. Plot Results Side-by-Side
    plt.figure(figsize=(12, 6))

    plt.subplot(1, 2, 1)
    plt.imshow(background_rgb)
    plt.title("Hard Pixel Overlay")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(blended_rgb)
    plt.title("Alpha Blended Overlay (60/40)")
    plt.axis("off")

    plt.show()
```

```python
#EXP3-Convert the given images to a binary image using different thresholding techniques
import cv2
import numpy as np
from matplotlib import pyplot as plt

# 1. Load the image
img = cv2.imread('/content/fingerprint.jfif', 0)

# --- NEW: SHARPENING STEP ---
# Define a sharpening kernel
# This kernel enhances the center pixel and subtracts the surrounding pixels
kernel = np.array([[ 0, -1,  0],
                   [-1,  5, -1],
                   [ 0, -1,  0]])

# Apply the sharpening filter
sharpened = cv2.filter2D(img, -1, kernel)
# ----------------------------

# 2. Perform Thresholding on the SHARPENED image
# This helps Otsu's find a cleaner separation between ridges and background
ret, thresh_bin = cv2.threshold(sharpened, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

# Use that 'ret' value for all other types
_, thresh_bin_inv = cv2.threshold(sharpened, ret, 255, cv2.THRESH_BINARY_INV)
_, thresh_trunc   = cv2.threshold(sharpened, ret, 255, cv2.THRESH_TRUNC)
_, thresh_tozero  = cv2.threshold(sharpened, ret, 255, cv2.THRESH_TOZERO)
_, thresh_tozero_inv = cv2.threshold(sharpened, ret, 255, cv2.THRESH_TOZERO_INV)

# 3. Display the results
# I've added 'Sharpened' to the list so you can see the difference from 'Original'
titles = ['Original', 'Sharpened', 'Binary (on Sharpened)', 'Binary Inv', 'Truncate', 'ToZero Inv']
images = [img, sharpened, thresh_bin, thresh_bin_inv, thresh_trunc, thresh_tozero_inv]

plt.figure(figsize=(15, 10))
for i in range(6):
    plt.subplot(2, 3, i+1)
    # vmin=0, vmax=255 prevents visual wash-out
    plt.imshow(images[i], cmap='gray', vmin=0, vmax=255)
    if i == 0 or i == 1:
        plt.title(titles[i])
    else:
        plt.title(f"{titles[i]} (T={int(ret)})")
    plt.xticks([]), plt.yticks([])

plt.tight_layout()
plt.show()
```
